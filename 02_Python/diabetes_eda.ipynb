# Data Cleaning and Validation

```python
# Check the missing values in the data
df_clean.isnull().sum()
```

```python
# Check the duplicate rows in the data
df_clean.duplicated().sum()
```

```python
# Check basic Statistics of the data (Numerical column)
df_clean.describe()
```

```python
# Check unique values in string column 
categorical_cols = df_clean.select_dtypes(include= "str").columns

for cols in categorical_cols:
    print(df_clean[cols].value_counts())
```

```python
# Check unique values in the numeric columns
numeric_cols = df_clean.select_dtypes(include= np.number).columns
for cols in numeric_cols:
    print(df_clean[cols].value_counts())
```

```python
# Check outliers in the data
for cols in numeric_cols:
    Q1 = df_clean[cols].quantile(0.25)
    Q3 = df_clean[cols].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR 
    upper_bound = Q3 + 1.5 * IQR  
    outliers = df_clean[(df_clean[cols] < lower_bound) | (df_clean[cols] > upper_bound)]
    print(f"{cols}: {len(outliers)} outliers")
```

```python
# Boxplot
for cols in numeric_cols:
    plt.figure(figsize=(6,3))
    plt.boxplot(df_clean[cols], vert= False)
    plt.title(cols)
    plt.show()
```
